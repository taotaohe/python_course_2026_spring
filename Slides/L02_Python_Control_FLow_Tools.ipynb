## 2. 判断语句 (if...else...)
Python条件语句是通过一条或多条语句的执行结果（`True` 或者 `False`）来决定执行的代码块。

* `if`、`elif`、`else` 语句末尾不要忘记写**冒号 `:`** 
* **注意缩进！** Python 是非常依赖缩进来区分代码块的语言。

````python
# 例子：判断输入值跟40的关系
inputv = 32

if inputv == 40:
    print('你输入的数值等于40！')
elif inputv < 40:
    print('你输入的数值小于40！')
else: 
    print('你输入的数值大于40！')

print('这一行命令每次都要运行，因为它没有缩进!')
````

## 3. 循环 (Loop)
程序在一般情况下是按顺序执行的，但是编程语言提供了各种控制结构，允许更复杂的执行路径。

循环语句允许我们执行一个语句或语句组多次。

![loop](figures/loop.png)

#### 循环的三种类型
- **while 循环**：在给定的判断条件为 True 时执行循环体，否则退出循环体。
- **for 循环**：重复执行语句，通常用于遍历序列。
- **嵌套循环**：你可以在 while 循环体中嵌套 for 循环，反之亦然。

### 3.1 while 循环

````python
# 输出10以内的所有奇数
a = 1

while a < 10:
    print(a)
    a = a + 2
````

````python
# 使用 while 打印出列表内的所有元素
mylist = [1, 1, 2, 3, 5, 8]

i = 0
while i < len(mylist):
    print(mylist[i])
    i = i + 1
````

#### continue 的用法

````python
# 输出10（包含）以内的所有偶数
i = -1

while i < 10:  
    i = i + 1
    if i%2 != 0: # 奇数时跳过当次循环后续的代码
        continue
    print(i)  # 输出偶数
````

#### break 的用法

````python
i = 0
while True:         # 循环条件为True，如果不中断将永远执行下去
    print(i)        # 输出1~10
    i += 2
    if i > 10:      # 当i大于10时跳出（结束）整个循环
        break
````

### 3.2 for 循环

````python
# 打印list
data_list = [1, 1, 2, 3, 5, 8]

# While 循环的方式
i = 0
while i < len(data_list):
    print(data_list[i])
    i = i + 1
    
# 上面的列表遍历非常常见，而 for 循环能让我们写得更快更简洁。
print("\nLet's try that again")
# 通过 range() 函数生成索引
for i in range(0, len(data_list)):
    print(data_list[i])

print("\nOne more time:")
# 或者更优雅的 Pythonic 写法，直接遍历元素本身！
for ee in data_list:
    print(ee)
````

## 4. 函数 (Function)
函数是组织好的，可重复使用的，用来实现单一，或相关联功能的代码段。<br>
函数能提高应用的模块性，和代码的重复利用率。你已经知道 Python 提供了许多内建函数，比如 `print()`。但你也可以自己创建函数，这被叫做用户自定义函数。

你可以定义一个由自己想要功能的函数，以下是简单的规则：

1. 函数代码块以 `def` 关键词开头，后接函数标识符名称和圆括号 `()`。<br>
2. 任何传入参数和自变量必须放在圆括号中间。圆括号之间可以用于定义参数。<br>
3. 函数的第一行语句可以选择性地使用文档字串（Doc strings）用于存放函数说明。<br>
4. 函数内容以冒号起始，并且**严格缩进**。<br>
5. 使用 `return` 结束函数，选择性地返回一个值给调用方。

````python
def print_you(string):
    string = string + 1
    print(string)
    return  # End with the "return" keyword
````

````python
print_you(1)
````

````python
# 定义另一个函数
def step(x):
    if x < 0:
        return -1
    elif x > 0:
        return 1
````

````python
# 调用函数
print(step(-5))
print(step(10))
````

````python
# 对于上面step函数，如果当x=0，会输出什么结果?
print(step(0))

# 如果你忘了某个条件，Python会自动增加一个“return None”的语句
````

````python
# 修正这个bug
def step_v2(x):
    if x < 0:
        return -1
    elif x >= 0:
        return 1
    
print(step_v2(0))
````

### 4.1 函数参数
<font color='tomato'><b>形式参数 (形参)</b></font> 与 <font color='tomato'><b>实际参数 (实参)</b></font>

- <font color='tomato'><b>形式参数</b></font>：定义或创建函数时，函数名称后面括号中的参数是形式参数，用于接收调用函数时传递的值。
- <font color='tomato'><b>实际参数</b></font>：调用函数时，函数名称后面括号中的参数是实际参数，也就是函数调用者提供给函数的参数。

````python
# a, b 是形参
def add_numbers(a,b):
    sum = a + b
    return sum
````

````python
# 5, 3 是实参
add_numbers(5,3)
````

函数传参时，我们可以使用不同类型的参数：

<font color='dodgerblue'><b>必备参数</b></font>

须以正确的顺序传入函数。调用时的数量必须和声明时的一样。

````python
# 调用printme()函数，你必须传入一个参数，不然会出现语法错误：
def printme(string):
    print(string)
    return
 
# printme()  # 如果去掉注释，这行代码将报错
````

````python
printme('hello')
````

<font color='dodgerblue'><b>关键字参数</b></font>

函数调用时使用关键字参数来确定传入的参数值。
使用关键字参数允许函数调用时参数的顺序与声明时不一致，因为 Python 解释器能够用参数名匹配参数值。

````python
def printinfo(name, age):
    print("Name: ", name)
    print("Age ", age)
    return
````

````python
print('Jone', 50)
````

````python
# 调用printinfo函数，顺序反了也没事
printinfo(age=50, name="miki")
````

````python
printinfo(name="miki", age=50)
````

<font color='dodgerblue'><b>默认参数</b></font>

调用函数时，默认参数的值如果没有传入，则会输出默认值。

````python
def printinfo(name, age=30):
    print("Name: ", name)
    print("Age ", age)
    return
 
# 由于第二个参数有默认值，我们可以只传一个参数
printinfo("miki")
````

<font color='dodgerblue'><b>不定长参数</b></font>

你可能需要一个函数能处理比当初声明时更多的参数。这些参数叫做不定长参数，声明时会在参数名前加星号 `*`。

````python
def printinfo(arg1, *vartuple):
    print(arg1)
    for var in vartuple:
        print(var*2)
    return
 
# 调用printinfo 函数
printinfo(10)
printinfo(70, 60, 50, 40)
````

### 4.2 全局变量和局部变量
定义在函数**内部**的变量称为局部变量（Local Variable），它的作用域范围仅限函数内。

函数的**外部**定义变量，称为全局变量（Global Variable）。

全局变量的作用域是整个程序，即全局变量既可以在各个函数的外部使用，也可以在各函数内部使用。

````python
product = 100 # 全局变量 product

def demo():
    n = 10    # 局部变量 n
    print(product)
    return n
````

````python
# 变量 product 在函数外定义，是全局变量，程序中任何位置使用它都可以。
demo()
````

````python
# 变量 n 在函数 demo 中定义，是局部变量，作用域范围在函数 demo() 内。
print(n) # 如果运行这行代码会报错，因为 n 是局部变量
````

如果要把函数内部的变量变成全局变量，则必须在该函数内将该变量以 `global` 关键字声明。

````python
a = 2 # 全局变量

def test():
    global a   # 声明 a 是全局变量，在该函数内部对a的操作就是对全局变量a的操作
    a = 1      # 如果没有上一行代码，默认a就是函数内部的局部变量
    print(a)
    # 函数结束，局部变量的生命周期结束
````

````python
test()

# 输出全局变量a
print(a)
````
