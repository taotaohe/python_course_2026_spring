### 1.2 打开和关闭文件

#### open() 函数

在对文件进行读写之前，需要先使用 Python 内置的 `open()` 函数打开文件，得到一个文件对象（file object），然后才能继续操作。

基本语法如下：

```python
file_object = open(file_name, mode, buffering)
```

其中最常见的参数有：

- `file_name`：要打开的文件名
- `mode`：打开模式，例如只读、写入、追加等
- `buffering`：缓冲区设置，初学阶段通常可以忽略，使用默认值即可

常见文件模式示例：

- `r`：只读
- `w`：写入；如果文件已存在，会覆盖原内容
- `a`：追加；在原文件末尾继续写入
- `r+`：可读可写
- `a+`：可读可追加

下面先创建一个文本文件，并向其中写入两行内容：

````python
file = open('my_first_file.txt', 'w')
````

````python
file.write('This is the first line.\n')
````

````python
file.write('This is the second line.\n')
````

````python
file.close()
````

## 3 面向对象编程入门（OOP）

### 3.1 什么是类（Class）和对象（Object）

类（Class）可以理解为一类事物的“模板”或“模具”；  
对象（Object）则是根据这个模板创建出来的具体实例。

例如：

- “人”可以看作一个类
- “张三”“李四”都是这个类创建出来的对象

面向对象编程（Object-Oriented Programming, OOP）是一种组织程序的方式。  
它的核心思想是：**把数据和操作数据的行为组织到同一个对象中**。

先别急着背定义，我们先通过一个直观例子来理解。

### 例子：洗衣服这件事

如果用**面向过程**的思路，我们会把问题拆成一系列动作：

1. 加洗衣粉  
2. 加水  
3. 洗衣服  
4. 清洗  
5. 烘干  

如果用**面向对象**的思路，我们会先找到“谁来做这些事”：

- <font color='blue'><b>洗衣机</b></font>：负责洗衣、清洗、烘干...(方法)
- <font color='limegreen'><b>人</b></font>：负责加洗衣粉、加水...(方法)

然后执行：
- <font color='limegreen'><b>人</b></font>.加洗衣粉

- <font color='limegreen'><b>人</b></font>.加水

- <font color='blue'><b>洗衣机</b></font>.洗衣服

- <font color='blue'><b>洗衣机</b></font>.清洗

- <font color='blue'><b>洗衣机</b></font>.烘干

也就是说，面向对象更强调“对象”和“对象的方法”。

### 再举一个例子：对象和类的关系是什么？
类是用来创建对象的模具，对象是类的实例（Instance）。

在海滩上一群小孩子在堆沙堡，他们先把沙子装进不同的塑料模具里，然后再把成型的沙子倒出来，堆成一个个漂亮的沙堡。一个制造沙堡柱子的塑料模具可以制造出无数个带有相同花纹的沙柱。在工业制造上，使用模具来铸造机壳和零件更是司空见惯。所以在计算机编程语言中引入“模具”来制造用来编程的“零件”也就顺其自然了。

面向对象编程中，将变量（属性）和函数（方法）进一步封装成类，类才是程序的基本元素。

### 3.2 创建一个最简单的类

下面我们先定义一个简单的 `Washer` 类，让它具备“洗衣服”的能力。

````python
# step 1：定义洗衣机类
class Washer:
    def wash(self):
        print("正在洗衣服……")

# step 2：通过类创建对象
my_haier_washer = Washer()

# step 3：调用对象的方法
my_haier_washer.wash()
````

### 3.3 属性（attribute）和方法（method）

每个类都可以包含：

- **属性（attribute）**：对象“有什么”
- **方法（method）**：对象“能做什么”

下面我们来创建一个 `Person` 类。

````python
# step 1：定义人的类
class Person:
    # attribute 属性
    name = 'William'
    age = 45

    # method 方法
    def greet(self):
        print("Hi, my name is " + self.name)

# step 2：通过类创建对象
p1 = Person()

# step 3：调用方法
p1.greet()
````

在 `greet(self)` 中，`self` 表示“当前这个对象自己”。  
因此 `self.name` 指向的就是当前实例的 `name` 属性。

对象创建好以后，我们也可以对它的属性进行修改。

````python
# 修改对象属性
p1.age = 20
p1.age
````

````python
# 修改的只是 p1 这个实例，不会影响其他实例
p2 = Person()
p2.age
````

````python
# 动态增加一个属性
p1.gender = 'male'
p1.gender
````

````python
# 删除对象属性
del p1.gender
hasattr(p1, 'gender')
````

### 3.4 `__init__()`：初始化对象

`__init__()` 是 Python 中的一个特殊方法。  
当我们创建对象时，它会自动执行，通常用来给对象设置初始属性。

````python
class Person:
    # 初始化方法
    def __init__(self):
        self.name = 'Alice'

    def greet(self):
        print("Hi, my name is " + self.name)

p1 = Person()
p1.greet()
````

我们也可以在 `__init__()` 中加入参数，让不同对象在创建时就拥有不同的初始值。

````python
class Person:
    def __init__(self, init_name):
        self.name = init_name

    def greet(self):
        print("Hi, my name is " + self.name)

p1 = Person("Bob")
p1.greet()
````

### 3.5 继承（Inheritance）和多态（Polymorphism）

面向对象编程中，常见的三个核心概念是：

- **封装（Encapsulation）**
- **继承（Inheritance）**
- **多态（Polymorphism）**

其中，继承和多态最容易通过例子理解。

#### 继承

当我们已经写好了一个类，又想创建一个“很像它”的新类时，  
就可以让新类继承旧类的属性和方法，而不必从头重写。

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print('Hello, I am %s.' % self.name)
````

现在我们想定义一个 `Dog` 类。最笨的方法当然是重新写一遍，但更好的方式是直接继承 `Animal`。

````python
class Dog(Animal):
    def greet(self):
        print('WangWang..., I am %s.' % self.name)
````

子类 `Dog` 会自动拥有父类 `Animal` 的属性，同时也可以改写自己的方法。

````python
animal = Animal('animal')
animal.greet()

dog = Dog('doggy')
dog.greet()
````

#### 多态

多态指的是：**同一个接口，面对不同对象时，可以表现出不同的行为**。

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print(f'Hello, I am {self.name}.')

class Dog(Animal):
    def greet(self):
        print(f'WangWang..., I am {self.name}.')

class Cat(Animal):
    def greet(self):
        print(f'MiaoMiao..., I am {self.name}.')

# hello 是统一的接口
def hello(animal):
    animal.greet()
````

````python
animal = Animal('animal')
dog = Dog('dog')
cat = Cat('cat')

hello(animal)
hello(dog)
hello(cat)
````
