## 2 Pandas 的数据结构介绍

要使用 Pandas，首先需要熟悉它的两个核心数据结构：

- **Series**：带索引的一维数组
- **DataFrame**：带行索引和列索引的二维表格

可以把 Series 理解为“一列数据”，把 DataFrame 理解为“一张表”。


### 2.1 Series

Series 是一种类似一维数组的对象，由两部分组成：

- 一组数据值
- 一组与数据值对应的标签，也就是**索引（index）**

如果没有显式指定索引，Pandas 会自动创建从 `0` 到 `N-1` 的整数索引。


````python
obj = pd.Series([4, 7, -5, 3])
obj
````

Series 的显示结果通常左边是索引，右边是数据值。  
我们可以通过 `.values` 和 `.index` 属性分别查看数据值和索引对象。


````python
obj.values
````

````python
obj.index
````

在实际数据分析中，我们通常会给 Series 指定有意义的索引标签，这样可以更清楚地表示每个数据点的含义。

````python
obj2 = pd.Series([4, 7, -5, 3], index=["d", "b", "a", "c"])
obj2
````

````python
obj2.index
````

与普通 NumPy 数组相比，Series 不仅可以按位置选取数据，还可以通过**索引标签**选取单个或多个值。

````python
obj2["a"]
````

````python
obj2["d"] = 6
obj2[["c", "a", "d"]]
````

`["c", "a", "d"]` 是一个索引标签列表。  
即使这些标签是字符串，也可以像使用 NumPy 数组一样进行筛选、数学运算和函数调用。Series 会在运算过程中保留索引信息。


````python
obj2[obj2 > 0]
````

````python
obj2 * 2
````

````python
np.exp(obj2)
````

Series 也可以看作一个**定长、有序的字典**：索引是键，数据值是值。  
因此，很多原本可以使用字典的场景，也可以使用 Series。


````python
"b" in obj2
````

````python
"e" in obj2
````

如果数据已经存放在 Python 字典中，也可以直接用这个字典创建 Series。

````python
sdata = {"Ohio": 35000, "Texas": 71000, "Oregon": 16000, "Utah": 5000}
obj3 = pd.Series(sdata)
obj3
````

````python
obj3.to_dict()
````

当使用字典创建 Series 时，默认索引来自字典的键。  
如果额外传入 `index` 参数，Pandas 会按照指定顺序重排数据；找不到的标签会产生缺失值。


````python
states = ["California", "Ohio", "Oregon", "Texas"]
obj4 = pd.Series(sdata, index=states)
obj4
````

Pandas 提供了 `isna()` 和 `notna()` 来检测缺失数据。  
在数据清洗中，缺失值处理是非常常见的步骤。


````python
pd.isna(obj4)
````

````python
pd.notna(obj4)
````

Series 对象本身也有对应的实例方法，例如 `.isna()`。

````python
obj4.isna()
````

Series 的一个重要特性是：**按索引标签自动对齐数据**。  
当两个 Series 做运算时，Pandas 会先按照索引匹配，再进行计算。


````python
obj3
````

````python
obj4
````

````python
obj3 + obj4
````

### 2.2 DataFrame

DataFrame 是 Pandas 中最常用的数据结构，可以理解为一张带标签的二维表格。

它具有以下特点：

- 每一列可以有不同的数据类型，例如数值、字符串、布尔值等
- 同时具有行索引和列索引
- 可以看作由多个 Series 按列组成的字典
- 非常适合处理表格型数据，例如 CSV、Excel、数据库查询结果等


创建 DataFrame 的方式有很多。最常用的方法之一是传入一个由等长列表或 NumPy 数组组成的字典。

````python
data = {"state": ["Ohio", "Ohio", "Ohio", "Nevada", "Nevada", "Nevada"],
        "year": [2000, 2001, 2002, 2001, 2002, 2003],
        "pop": [1.5, 1.7, 3.6, 2.4, 2.9, 3.2]}

frame = pd.DataFrame(data)
frame
````

对于较大的 DataFrame，可以使用 `.head()` 查看前几行，使用 `.tail()` 查看后几行。

````python
frame.head()
````

````python
# 选最后五行
frame.tail()
````

如果指定 `columns` 参数，DataFrame 的列会按照指定顺序排列。

````python
pd.DataFrame(data, columns=["year", "state", "pop"])
````

如果指定的列在原始数据中不存在，Pandas 会创建该列，并用缺失值填充。

````python
frame2 = pd.DataFrame(data, columns=['year', 'state', 'pop', 'debt'],
                      index=['one', 'two', 'three', 'four', 'five', 'six'])
frame2
````

可以像访问字典一样获取 DataFrame 的某一列，也可以在列名符合变量命名规则时使用属性方式访问。

````python
frame2.year
````

````python
frame2.loc['three']  # 按行标签选取
````

````python
frame2.iloc[2]  # 按整数位置选取
````

DataFrame 的列可以通过赋值方式修改。  
既可以赋一个标量值，也可以赋一个长度匹配的数组。


````python
frame2["debt"] = 16.5
frame2
````

````python
frame2["debt"] = np.arange(6.)
frame2
````

给某一列赋列表或数组时，长度必须和 DataFrame 的行数一致。  
如果赋值的是 Series，Pandas 会按照索引标签自动对齐，未匹配的位置会填入缺失值。


````python
val = pd.Series([-1.2, -1.5, -1.7], index=["two", "four", "five"])
frame2["debt"] = val
frame2
````

给一个不存在的列赋值，会创建新列。  
也可以使用 `del` 删除已有列。


````python
frame2["eastern"] = frame2["state"] == "Ohio"
frame2
````

## 3 索引对象

Pandas 的索引对象负责管理轴标签和其他元数据，例如索引名称。  
构建 Series 或 DataFrame 时，传入的标签序列通常会被转换成 Index 对象。


````python
obj = pd.Series(np.arange(3), index=["a", "b", "c"])
idx = obj.index
idx
````

Index 对象是**不可变的**，不能直接修改其中某一个位置的值。  
这种设计可以让 Index 在多个数据结构之间安全共享。


````python
try:
    idx[1] = "d"
except TypeError as err:
    print("Index 对象不可变，不能直接修改单个位置的值。")
    print("错误信息：", err)
````

### 4.1 重新索引

`reindex()` 的作用是按照新的索引创建一个新对象。  
如果新索引中包含原对象没有的标签，对应位置会产生缺失值。


````python
obj = pd.Series([4.5, 7.2, -5.3, 3.6], index=["d", "b", "a", "c"])
obj
````

使用 `reindex()` 可以按照新的索引顺序重排 Series。  
如果某个索引值不存在，就会引入缺失值。


````python
obj2 = obj.reindex(["a", "b", "c", "d", "e"])
obj2
````

对于时间序列等有序数据，重新索引时经常需要做填充。  
`method="ffill"` 表示用前一个有效值向前填充。


````python
obj3 = pd.Series(["blue", "purple", "yellow"], index=[0, 2, 4])
obj3
````

````python
obj3.reindex(np.arange(6), method="ffill")
````

对于 DataFrame，`reindex()` 既可以重新索引行，也可以重新索引列。  
如果只传入一个序列，默认会重新索引行。


````python
frame = pd.DataFrame(np.arange(9).reshape((3, 3)),
                     index=["a", "c", "d"],
                     columns=["Ohio", "Texas", "California"])
frame
````

````python
frame2 = frame.reindex(index=["a", "b", "c", "d"])
frame2
````

````python
states = ["Texas", "Utah", "California"]
frame.reindex(columns=states)
````

````python
obj = pd.Series(np.arange(5.), index=["a", "b", "c", "d", "e"])
obj
````

````python
new_obj = obj.drop("c")
new_obj
````

````python
obj.drop(["d", "c"])
````

传入 `index` 参数或默认按行标签删除，可以移除指定行。

````python
data.drop(index=["Colorado", "Ohio"])
````

### 4.4 使用 `loc` 和 `iloc` 进行选取

DataFrame 选取数据时，最推荐使用两个明确的索引器：

- `.iloc[]`：按**整数位置**选取

````python
data = pd.DataFrame(np.arange(16).reshape((4, 4)),
                    index=["Ohio", "Colorado", "Utah", "New York"],
                    columns=["one", "two", "three", "four"])
data
````

````python
data.loc["Colorado"]
````

````python
data.loc[["Colorado", "New York"]]
````

````python
data.loc["Colorado", ["two", "three"]]
````

下面使用 `.iloc[]` 按整数位置进行选取。

````python
data.iloc[2]
````

````python
data.iloc[[2, 1]]
````

````python
data.iloc[2, [3, 0, 1]]
````

````python
data.iloc[[1, 2], [3, 0, 1]]
````

这两个索引器也可以配合切片、列表和条件过滤使用。

````python
data.loc[:"Utah", "two"]
````

````python
data.iloc[:, :3][data.three > 5]
````

### <font color='cornflowerblue'><b>思考题 1</b></font>

`loc` 和 `iloc` 有什么区别？为什么在 Pandas 中建议显式使用它们，而不是直接写 `obj[1]`？

#### 参考答案

<details>
<summary>点击查看解释</summary>

- `loc` 按**标签**选取数据，例如 `df.loc["row_name", "col_name"]`
- `iloc` 按**整数位置**选取数据，例如 `df.iloc[0, 1]`
- 直接写 `obj[1]` 时，`1` 可能被理解为标签，也可能被理解为位置，容易产生歧义
- 显式使用 `loc` 或 `iloc` 可以让代码更清晰、更稳定

</details>


### Pandas 常用选取方式小结

| 操作 | 示例 | 说明 |
|---|---|---|
| 选取单列 | `df["col"]` | 返回一个 Series |
| 选取多列 | `df[["col1", "col2"]]` | 返回一个 DataFrame |

### 4.5 算术运算和数据对齐

Pandas 的重要特性之一是**自动数据对齐**。  
当两个 Series 或 DataFrame 做运算时，Pandas 会根据索引标签进行匹配；不匹配的位置会产生缺失值。

这类似于数据库中的“外连接”思想。


````python
s1 = pd.Series([7.3, -2.5, 3.4, 1.5], index=["a", "c", "d", "e"])
s1
````

````python
s2 = pd.Series([-2.1, 3.6, -1.5, 4, 3.1], index=["a", "c", "e", "f", "g"])
s2
````

````python
s1 + s2
````

````python
df1 = pd.DataFrame(np.arange(9.).reshape((3, 3)), columns=list("bcd"),
                   index=["Ohio", "Texas", "Colorado"])
df2 = pd.DataFrame(np.arange(12.).reshape((4, 3)), columns=list("bde"),
                   index=["Utah", "Ohio", "Texas", "Oregon"])
````

````python
df1
````

````python
df2
````

````python
df1 + df2
````

如果两个 DataFrame 没有共同的行标签或列标签，相加后的结果会全部是缺失值。

````python
df1 = pd.DataFrame({"A": [1, 2]})
df2 = pd.DataFrame({"B": [3, 4]})
df1
````

````python
df2
````

````python
df1 + df2
````

### 4.6 在算术方法中填充值

在对不同索引的对象进行算术运算时，可以通过 `fill_value` 指定缺失位置的填充值。  
例如，将缺失位置当作 0 再进行加法。


````python
df1 = pd.DataFrame(np.arange(12.).reshape((3, 4)), columns=list("abcd"))
df2 = pd.DataFrame(np.arange(20.).reshape((4, 5)), columns=list("abcde"))
df2.loc[1, "b"] = np.nan
````

````python
df1
````

````python
df2
````

````python
df1 + df2
````

使用 `df1.add(df2, fill_value=0)` 时，Pandas 会先把只有一边存在的位置补为 0，然后再相加。

````python
df1.add(df2, fill_value=0)
````

### <font color='limegreen'><b>技巧与提示</b></font>

Pandas 的算术运算会先根据索引和列名自动对齐数据。  
如果不希望因为标签不匹配而产生大量缺失值，可以考虑：

- 使用 `fill_value` 指定缺失位置的默认值
- 先用 `reindex()` 统一索引和列
- 在运算前检查两个对象的 `.index` 和 `.columns`


````python
df1.add(df2)
````

类似地，在重新索引时，也可以用 `fill_value` 指定新位置的填充值。

````python
df1.reindex(columns=df2.columns, fill_value=0)
````

### 4.7 DataFrame 和 Series 之间的运算

DataFrame 和 Series 之间的运算类似于 NumPy 中的广播机制。  
先来看一个 NumPy 数组的例子：二维数组减去第一行。


````python
arr = np.arange(12.).reshape((3, 4))
arr
````

````python
arr[0]
````

````python
arr - arr[0]
````

当二维数组减去其中一行时，这一行会被广播到每一行。  
DataFrame 和 Series 之间的运算也有类似规则。


````python
frame = pd.DataFrame(np.arange(12.).reshape((4, 3)),
                     columns=list("bde"),
                     index=["Utah", "Ohio", "Texas", "Oregon"])
frame
````

````python
series = frame.iloc[0]
series
````

默认情况下，DataFrame 和 Series 的算术运算会把 Series 的索引匹配到 DataFrame 的列，然后沿行方向广播。

````python
frame - series
````

如果某个标签在 DataFrame 的列或 Series 的索引中找不到，Pandas 会取两边标签的并集，未匹配的位置产生缺失值。

````python
series2 = pd.Series(np.arange(3), index=["b", "e", "f"])
series2
````

````python
frame + series2
````

如果希望按行索引匹配，并在列方向上广播，需要使用算术方法并指定 `axis`。

````python
frame
````

````python
series3 = frame["d"]
series3
````

````python
frame.sub(series3, axis="index")
````

这里的 `axis="index"` 表示按照 DataFrame 的行索引进行匹配。

### 4.8 函数应用和映射

NumPy 的通用函数（ufunc）可以直接作用于 Pandas 对象。  
此外，Pandas 还提供了 `apply()` 和 `map()` 等方法，用于更灵活地应用自定义函数。


````python
frame = pd.DataFrame(np.random.standard_normal((4, 3)),
                     columns=list("bde"),
                     index=["Utah", "Ohio", "Texas", "Oregon"])
frame
````

````python
np.abs(frame)
````

另一个常见操作是把函数应用到每一列或每一行上。  
DataFrame 的 `apply()` 方法可以完成这类任务。


````python
def f1(x):
    return x.max() - x.min()

frame.apply(f1)
````

下面的函数计算一个 Series 的最大值和最小值之差。  
默认情况下，`apply()` 会对每一列执行函数，并返回一个以列名为索引的 Series。


````python
frame.apply(f1, axis="columns")
````

很多常见统计功能已经被实现为 DataFrame 方法，例如 `.sum()`、`.mean()` 等，因此不一定需要自己写 `apply()`。  
如果传入的函数返回多个值，也可以返回一个 Series。


````python
def f2(x):
    return pd.Series([x.min(), x.max()], index=["min", "max"])
frame.apply(f2)
````

元素级函数可以使用 `map()`。  
例如，把每个浮点数格式化为保留两位小数的字符串。


````python
def my_format(x):
    return f"{x:.2f}"

frame.map(my_format)
````

Series 也有 `map()` 方法，常用于元素级转换。

````python
frame["e"].map(my_format)
````

### 4.9 排序和排名

排序是数据分析中非常常见的操作。  
Pandas 中常见的排序方法包括：

- `sort_index()`：按索引排序
- `sort_values()`：按值排序
- `rank()`：计算排名


````python
obj = pd.Series(np.arange(4), index=["d", "a", "b", "c"])
obj
````

````python
obj.sort_index()
````

对于 DataFrame，可以根据行索引或列索引排序。

````python
frame = pd.DataFrame(np.arange(8).reshape((2, 4)),
                     index=["three", "one"],
                     columns=["d", "a", "b", "c"])
frame
````

````python
frame.sort_index(axis="columns")
````

默认情况下按升序排序，也可以通过 `ascending=False` 设置为降序。

````python
frame.sort_index(axis="columns", ascending=False)
````

如果要按 Series 的值排序，可以使用 `sort_values()`。

````python
obj = pd.Series([4, 7, -3, 2])
obj.sort_values()
````

排序时，缺失值默认会被放在末尾。

````python
obj = pd.Series([4, np.nan, 7, np.nan, -3, 2])
obj.sort_values()
````

对 DataFrame 排序时，常常需要按照某一列或多列的值进行排序。

````python
frame = pd.DataFrame({"b": [4, 7, -3, 2], "a": [0, 1, 0, 1]})
frame.sort_values(by="b")
````

按照多个列排序时，将列名列表传给 `by` 参数即可。

````python
frame.sort_values(by=["a", "b"])
````

`rank()` 用于计算排名。  
默认情况下，如果出现并列值，会给这些值分配平均排名。


````python
obj = pd.Series([7, -5, 7, 4, 2, 0, 4])
obj.rank()
````

也可以使用 `method="first"`，按照数据在原始序列中出现的顺序打破并列关系。

````python
obj.rank(method="first")
````

也可以通过 `ascending=False` 计算降序排名。

````python
obj.rank(ascending=False)
````

````python
frame = pd.DataFrame({"b": [4.3, 7, -3, 2], "a": [0, 1, 0, 1],
                      "c": [-2, 5, 8, -2.5]})
frame
````

````python
frame.rank(axis="columns")
````

### 4.10 带有重复标签的轴索引

前面的例子大多使用唯一索引。  
实际上，Pandas 允许索引标签重复，但重复标签会让数据选取结果变得更复杂。


````python
obj = pd.Series(np.arange(5), index=["a", "a", "b", "b", "c"])
obj
````

````python
obj.index.is_unique
````

对于重复索引，如果某个标签对应多个值，选取结果会返回 Series；如果只对应一个值，则返回标量。

````python
obj["a"]
````

````python
obj["c"]
````

这会让代码逻辑变复杂，因为返回类型会根据标签是否重复而变化。  
DataFrame 的行索引也有类似行为。


````python
df = pd.DataFrame(np.random.standard_normal((5, 3)),
                  index=["a", "a", "b", "b", "c"])
df
````

````python
df.loc["b"]
````

````python
df.loc["c"]
````

## 5 汇总和计算描述统计

Pandas 提供了一组常用的数学和统计方法，用于快速汇总 Series 或 DataFrame。  
这些方法通常会自动跳过缺失值，适合用于数据分析初步探索。


````python
df = pd.DataFrame([[1.4, np.nan], [7.1, -4.5],
                   [np.nan, np.nan], [0.75, -1.3]],
                  index=["a", "b", "c", "d"],
                  columns=["one", "two"])
df
````

调用 DataFrame 的 `.sum()` 方法，默认会按列求和，返回一个 Series。

````python
df.sum()
````

传入 `axis="columns"` 或 `axis=1`，可以按行进行求和。

````python
df.sum(axis="columns")
````

缺失值默认会被自动排除。  
如果设置 `skipna=False`，只要参与计算的数据中存在缺失值，结果也会变成缺失值。


````python
df.sum(axis="index", skipna=False)
````

````python
df.sum(axis="columns", skipna=False)
````

````python
df.mean(axis="columns")
````

有些方法返回的是间接统计结果，例如 `.idxmin()` 和 `.idxmax()` 返回最小值或最大值所在的索引。

````python
df.idxmax()
````

另一些方法是累计型方法，例如 `.cumsum()` 会计算累计和。

````python
df.cumsum()
````

`describe()` 可以一次性生成多个描述统计指标，是快速了解数据分布的常用方法。

````python
df.describe()
````

对于非数值型数据，`describe()` 会返回另一类汇总信息，例如唯一值数量、出现最多的值及其频数。

````python
obj = pd.Series(["a", "a", "b", "c"] * 4)
obj.describe()
````

### `describe()` 常见输出说明

| 数据类型 | 常见统计量 | 说明 |
|---|---|---|
| 数值型 | `count`, `mean`, `std`, `min`, `25%`, `50%`, `75%`, `max` | 用于快速查看分布、集中趋势和离散程度 |
| 非数值型 | `count`, `unique`, `top`, `freq` | 用于查看类别数量、最常见类别及出现次数 |
