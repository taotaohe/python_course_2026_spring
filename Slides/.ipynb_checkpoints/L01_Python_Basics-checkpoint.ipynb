## 2 Python 基础

### 2.1 变量（Variable）

当我们在程序中创建一个数据时，计算机会在内存中为它分配存储空间。  
为了方便找到并使用这个数据，我们会给它取一个名字，这个名字就叫做 **变量**。

你可以把计算机内存想象成一个仓库，而变量就像贴在货物上的标签。  
通过这个标签，我们就能快速找到对应的数据。

简单来说：

- **变量**：数据的名字
- **值**：变量中保存的内容

变量中可以存储不同类型的数据，例如：

- 整数
- 小数
- 字符串
- ...

当变量保存的数据类型不同，计算机分配和处理数据的方式也会有所不同。

````python
age = 23
my_name = 'Tao He'
````

````python
print(age)
````

````python
type(age)
````

````python
# 连续变量命名
a = b = c = 1
````

````python
print(a)
````

````python
a, b, c = 1, 2, "john"
````

````python
print(b)
````

### <font color='limegreen'><b>技巧与提示</b></font>

在给变量命名时，需要注意以下规则：

1. **变量名只能包含字母、数字和下划线 `_`，且不能以数字开头**
   - 正确：`hello_world_1`
   - 错误：`$hello`、`*cool`

2. **变量名中不能包含空格**
   - 正确：`student_name`、`studentName`
   - 错误：`hello word`

3. **不要使用 Python 的关键字或内置函数名作为变量名**
   - 正确：`student`
   - 错误：`print`、`class`、`def`

4. **变量名应尽量简洁，并能表达实际含义**
   - 正确：`student_name`、`class_grade`
   - 错误：`sdf_23`

### 2.2 数字、字符串与索引（Numbers, Strings, & Index）

#### 2.2.1 数字（Numbers）

在 Python 中，我们可以直接对数字进行各种运算，例如加、减、乘、除、乘方、整除和取余。  
下面的代码将演示这些最常见的数值运算。

````python
a = 12
b = 2
````

````python
print(a + b)
print(a**b)    # 12 to the power of 2 (12 squared)
print(a/b)     # division always returns a floating point number
print(a//b)    # Floor Division (向下取整除) 只保留整数部分
print(a%b)     # remainder
````

````python
width  = 12
height = 10
print('area = ' + str(width*height))
````

#### 2.2.2 字符串（Strings）

字符串是由一系列字符组成的数据，通常用 **单引号 `' '`** 或 **双引号 `" "`** 括起来。  
当字符串中需要表示特殊字符时，可以使用反斜杠 `\` 进行转义。

````python
'spam eggs'
````

````python
'doesn\'t'
````

````python
"doesn't" # 或者用双引号替代
````

````python
s = 'First line.\nSecond line.'  # \n 换行符
print(s)
````

````python
# 字符串的开头加入r，让引号的内部字符串不转义
print('C:\some\name')  # \n 是换行符！
````

````python
print(r'C:\some\name')  # \n 这里不是换行符！
````

````python
# 有时字符串可能会横跨很多行，可以用三个双引号"""..."""。

print("""\
Usage: thingy [OPTIONS]
-h              Display this usage message
-H hostname     Hostname to connect to
""")
````

````python
# 加号（+）是字符串的连接符，星号（*）用来复制字符串
3 * 'un' + 'ium'
````

````python
# Python支持格式化字符串的输出，这有助于我们编写复杂的字符串：
print("%s was born in %d." %('Python', 1991))
````

````python
print("{} was born in {}.".format('Python', 1991))
````

### 一些常用的字符串函数

````python
name = "Hello world"
print(name.title()) # Hello World
print(name.upper()) # HELLO WORLD
print(name.lower()) # hello world
````

````python
str1 = "    Hello"
str2 = "  World"
# 用来消除空格的函数
print(str1.strip() + str2.strip()) # HelloWorld
````

#### 2.2.3 索引（Index）

字符串中的每个字符都有自己的位置编号，这个编号就叫做 **索引**。  
借助索引，我们可以访问字符串中的某一个字符，或截取其中的一部分内容。\
**注意：初始索引值是0而不是1！**

````python
word = 'Python'
print(word)
word[3]  
````

````python
word[-1]
````

````python
word[-2]
````

````python
word[0:3]  # 左闭右开区间
````

````python
word[:2] + word[3:]
````

### 2.3 列表（List）

列表是 Python 中最常用的数据类型之一。  
它由一组用逗号分隔的数据组成，并放在方括号 `[]` 中。

列表有两个重要特点：

- 列表中的元素可以是不同的数据类型
- 列表中的内容通常是可以修改的

创建列表时，只需要将多个数据项写在方括号中即可。

````python
list1 = [1, 2, 3, 4, 5]
list2 = ["a", "b", "c", "d"]
list3 = ['e', 'f', 100, 200]
````

````python
squares = [1, 4, 9, 16, 25]
print(squares)
````

````python
len(squares)
````

````python
squares[1:]
````

````python
squares + [36, 49, 64, 81, 100]
````

````python
# 为列表中每个元素都加上一个数
new_list = [x+1 for x in squares]
new_list
````

````python
squares.append('hello')
squares
````

````python
len(squares)
````

````python
x = [['a', 'b', 'c'], [1, 2, 3]]
````

````python
x[0][1]
````

### 2.4 元组（Tuple）

元组与列表非常相似，不同之处在于：**元组中的元素不能被修改**。  

- 列表使用方括号 `[]`
- 元组使用小括号 `()`

创建元组时，只需要把多个元素写在括号中，并用逗号隔开即可。

````python
tuple1 = (1, 2, 3, 4, 5)
tuple2 = ("a", "b", "c", "d")
tuple3 = ('e', 'f', 100, 200)
````

````python
tuple1[1]
````

````python
# 列表中的元素值可以修改
x[0] = 1
x
````

````python
# 元组中的元素值是不允许修改的
tuple1[0] = 100
````

````python
# 但我们可以对元组进行连接组合
tuple1+tuple2
````

### 2.5 布尔类型（Bool）

布尔类型只有两个取值：

- `True`
- `False`

它通常用于表示“是否成立”“是否正确”这类判断结果。

````python
string = "Hello Word"
print(string.isalnum()) # 查看是否所有字符都是数字
print(string.isupper()) # 查看字符串是否都是大写字母
````

````python
# 三种基本逻辑运算：and, or, not
True and False # False
True and True # True
False and True # False
False or True # True
False and False # False
not False # True
not True # False
````

### 2.6 字典（Dictionary）

字典是 Python 中另一种比较常用的数据结构，它用来存储 **键值对（key-value）** 数据。

字典的基本格式如下：

```python
d = {key1: value1, key2: value2}
```

其中：

- `key` 表示键
- `value` 表示与这个键对应的值

字典整体放在花括号 `{}` 中，不同的键值对之间用逗号分隔。

````python
# 键一般是唯一的，值不需要唯一。如果有两个重复的键值对，后面的会替换前面的。
mydict = {'a': 2, 'b': 3, 'c': '3'}
mydict
````

````python
mydict['c']
````

````python
# 值可以取任何数据类型，但键必须是不可变类型，如字符串，数字或元组，所以列表不行。
mynewdict = {'abc': 123, 98.6: 'char', ('a'): 456}
````

````python
mynewdict[98.6]
````

````python
mynewdict = {'abc': 123, 98.6: 'char', ['a']: 456}
````

````python
# 访问字典里的值
mydict1 = {'Name': 'Bob', 'Age': 10, 'Class': 'First'}
print(mydict1['Name'], 'is', mydict1['Age'], 'years old.')
````

````python
# 修改字典
mydict1['Age'] = 8 # 更新
mydict1['School'] = "BLCU" # 新增
````

````python
mydict1
````

````python
# 删除字典元素
del mydict1['Name']
mydict1
````

````python
mydict1.clear() # 清空字典里的所有条目
````

````python
mydict1
````

### 字典常用函数与方法

````python
mynewdict.items() # 以列表返回可遍历的(键, 值) 元组数组
````

````python
mynewdict.keys() # 以列表返回一个字典所有的键
````

````python
mynewdict.values() # 以列表返回字典中的所有值
````

### <font color='limegreen'><b>技巧与提示</b></font>

在选择数据结构时，可以先问自己三个问题：

1. 数据是否需要保持顺序？  
2. 数据是否允许修改？  
3. 我是按“位置”访问，还是按“名字/键”访问？  

一般来说：

- **列表（list）**：适合按顺序保存一组可修改的数据  
- **元组（tuple）**：适合保存不希望被修改的数据  
- **字典（dict）**：适合通过“键”快速找到对应的信息
